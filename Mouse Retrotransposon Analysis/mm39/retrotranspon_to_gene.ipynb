```python
import pandas as pd
import pyranges1 as pr
```

## Import mm39 annotations

```python
ann = pr.read_gtf("data/gencode.vM39.basic.annotation.gtf")
```

## Import mm39 rmsk

```python
rmsk_cols = [
    "bin", "swScore", "milliDiv", "milliDel",
    "milliIns", "Chromosome", "Start", "End", "genoLeft",
    "strand", "repName", "repClass", "repFamily",
    "repStart", "repEnd", "repLeft", "id"
]

df = pd.read_csv(
    "data/rmsk.txt",
    sep="\t",
    header=None,
    names=rmsk_cols
)

rmsk = pr.PyRanges(df)
```

## Filter retrotransposons and protein coding genes

```python
retrotransposons = rmsk[rmsk['repClass'].isin(['LINE', 'SINE', 'LTR'])]
```

```python
genes = ann[(ann['Feature'] == 'gene') & (ann['gene_type'] == 'protein_coding')]
```

## Nearest Retrotransposon to each gene

```python
genes.nearest_ranges(retrotransposons)
```

## k nearest distances between a retrotransposon and a selected gene
Returns all of the retrotransposons at each distance. So there will always be k distinct distances, but there could be more than one retrotransposon at each distance.

```python
gene = genes.sample(1)
k_nearest = gene.nearest_ranges(retrotransposons, k=5)
print(k_nearest.value_counts('Distance'))
k_nearest
```
